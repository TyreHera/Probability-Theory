# Парадокс "Спящей Красавицы"
---

## Условие парадокса
---
Представим следующий эксперимент:

1. В воскресенье вечером подбрасывают честную монету, Спящая Красавица в этот момент спит.
2. Если выпадает **орёл**, Красавицу будят в понедельник, задают ей вопрос, а затем дают препарат, стирающий воспоминания о пробуждении. Во вторник её уже не будят.
3. Если выпадает **решка**, Красавицу будят и в понедельник, и во вторник. После пробуждения в понедельник её воспоминания также стирают, поэтому во вторник она не помнит понедельник.
4. Каждый раз, когда Красавица просыпается, она не знает, какой сейчас день, и не помнит, просыпалась ли она раньше.

**Вопрос:** Красавица проснулась и видит, что находится в условиях эксперимента. Какова вероятность того, что монета выпала орлом?

Важно различать два вопроса:
- **До подбрасывания монеты:** вероятность орла равна \(1/2\). Это обычная вероятность исхода честной монеты.
- **В момент случайно выбранного пробуждения:** нужно учитывать, что при решке возникает два пробуждения, а при орле - только одно. Именно эту вероятность мы будем рассчитывать.
---

## Суть парадокса
---
Изначально кажется, что так как монета была честной, то вероятности выпадения решки и орла равные, поэтому ответ на задачу - **1/2**. Но это решение неверное, так как Красивицу будят в любой момент (либо в понедельник, либо во вторник), а если в воскресенье выпала решка, то ее будят **два** раза, поэтому правильным ответом на вопрос будет - **1/3**. Чтобы было понятнее, посмотрим на таблицу, описывающую множество всех элементарных исходов. Для случайно выбранного пробуждения возможны три равновероятных случая:

| Случай | Результат монеты | День пробуждения |
|---|---|---|
| 1 | Орёл | Понедельник |
| 2 | Решка | Понедельник |
| 3 | Решка | Вторник |

---

## Моделирование экспериментов
Сначала **N** раз подбрасываем монету в воскресенье и создаём список дней пробуждений. Должно получиться следущее:
* количество понедельников примерно в 2 раза больше количества вторников;
* количество решек должно быть в 2 раза больше количества орлов.

```python
N = 100_000
days_awake = []
for i in range(N):
    coin = "Орёл" if random.randint(0,1) else "Решка"
    if coin == "Орёл":
        days_awake.append(("Понедельник", "Орёл"))
    else:
        days_awake.append(("Понедельник", "Решка"))
        days_awake.append(("Вторник", "Решка"))

cnt_mondays = 0
cnt_tuesdays = 0
cnt_orels = 0
cnt_reshkas = 0
for day_awake in days_awake:
    if day_awake[1] == "Орёл":
        cnt_mondays += 1
        cnt_orels += 1
    else:
        cnt_reshkas += 1
        if day_awake[0] == "Понедельник":
            cnt_mondays += 1
        else:
            cnt_tuesdays += 1

print(f"Количество понедельников: {cnt_mondays}")
print(f"Количество вторников: {cnt_tuesdays}")
print(f"Количество орлов: {cnt_orels}")
print(f"Количество решек: {cnt_reshkas}\n")
print(f"Экспериментальная вероятность: {round(cnt_orels/len(days_awake), 4)}")
print(f"Теоретическая вероятность: {round(1/3, 4)}")
print(f"Абсолютная ошибка: {round(abs(1/3-cnt_orels/len(days_awake)), 4)}")
```
