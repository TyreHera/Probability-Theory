### Задача 1.17


```python
a = round(sc.comb(8, 7) * sc.comb(9, 2) / sc.comb(36, 7), 8)
print(f'а) {a:.7f}')
```

```python
b = round(4**7 * sc.comb(9, 7) / sc.comb(36, 7), 8)
print(f'б) {b:.5f}')
```

```python
c1 = sc.comb(4, 1) * sc.comb(4, 2) * sc.comb(4, 4) * sc.factorial(3)
c2 = sc.comb(4, 1) * sc.comb(4, 3) * sc.comb(4, 3) * sc.comb(3, 2)
c3 = sc.comb(4, 2) * sc.comb(4, 2) * sc.comb(4, 3) * sc.comb(3, 2)
c = round((c1 + c2 + c3) * sc.comb(9, 3) / sc.comb(36, 7), 6)
print(f'в) {c:.5f}')
```
