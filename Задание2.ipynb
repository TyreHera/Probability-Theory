#  Проверка связи частотного и классического определения вероятности с помощью моделирования
>В группе находится 23 студента. Какова вероятность, что хотя бы у двух студентов совпадает день рождения?

* Рассмотрим событие $A$ – хотя бы у двух студентов совпадает день рождения и пол.<br>Найдем вероятность противоположного события $\overline{A}$ – у всех студентов разные дни рождения.<br>Количество благоприятных исходов - $A_{365}^{23}$, общее количество исходов - ${365}^{23}$.<br>Тогда, $P(\overline{A}) = \frac{A_{365}^{23}}{{365}^{23}} \approx 0.4927$<br>И следовательно, $P(A) = 1 - P(\overline{A}) \approx 0.5073$

```python
import numpy as np
import matplotlib.pyplot as plt

def has_duplicate_birthday():
    birthdays = np.random.randint(1, 366, size=23)
    return len(set(birthdays)) < 23

N_values = np.array(range(10, 10002, 50))
P_A_values = []

for N in N_values:
    N_A = 0
    for _ in range(N):
        if has_duplicate_birthday():
            N_A += 1
    P_A_values.append(N_A / N)

print(f"Экспериментальная частота при N={N_values[-1]}: {round(P_A_values[-1], 4)}")
P_theoretical = [0.5073] * len(N_values)

plt.figure(figsize=(7, 3.5))
plt.plot(N_values, P_A_values, label='Частота (симуляция)', color='BLUE')
plt.plot(N_values, P_theoretical, label='Вероятность (~0.5073)', linestyle='--', color='RED')

plt.xlabel('N (Число экспериментов)')
plt.ylabel('P(A)')

plt.legend()
plt.tight_layout()
plt.show()
```
