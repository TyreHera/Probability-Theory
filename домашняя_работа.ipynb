### Задача 1.10
>Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к другой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

* Рассмотрим событие $A$ – из 3 карточек получится слово «ТОР»

```python
p_a = round(1 / sc.perm(6, 3), 4)
print(f'а) {p_a}')
```

* Рассмотрим событие $B$ – из 6 карточек получится слово «ТЕОРИЯ»

```python
p_b = round(1 / sc.factorial(6), 4)
print(f'б) {p_b}')
```

### Задача 1.16
>За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

* Рассмотрим событие $A$ – никакие два лица одного пола не сядут рядом

```python
p_a = round(2 / sc.comb(10, 5), 4)
print(f'а) {p_a}')
```

* Рассмотрим событие $B$ – супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар

```python
p_b = round(2**5 * sc.factorial(5) / sc.factorial(10), 4)
print(f'б) {p_b}')
```

### Задача 1.17
>В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

* Рассмотрим событие $A$ – пассажиры попали в два купе

```python
p_a = round(sc.comb(8, 7) * sc.comb(9, 2) / sc.comb(36, 7), 8)
print(f'а) {p_a:.7f}')
```

* Рассмотрим событие $B$ – пассажиры попали в семь купе

```python
p_b = round(4**7 * sc.comb(9, 7) / sc.comb(36, 7), 4)
print(f'б) {p_b}')
```

* Рассмотрим событие $C$ – пассажиры попали в три купе

```python
c1 = sc.comb(4, 1) * sc.comb(4, 2) * sc.comb(4, 4) * sc.factorial(3)
c2 = sc.comb(4, 1) * sc.comb(4, 3) * sc.comb(4, 3) * sc.comb(3, 2)
c3 = sc.comb(4, 2) * sc.comb(4, 2) * sc.comb(4, 3) * sc.comb(3, 2)
p_c = round((c1 + c2 + c3) * sc.comb(9, 3) / sc.comb(36, 7), 4)
print(f'в) {p_c}')
```

### Задача 1.43
>Наудачу взятый телефонный номер состоит из 5 цифр. Какова вероятность того, что в нем все цифры: а) различные; б) одинаковые; в) нечетные? Известно, что номер телефона не начинается с цифры ноль.

* Рассмотрим событие $A$ – все цифры различные

```python
p_a = round(9 * sc.perm(9, 4) / (9 * 10**4), 4)
print(f'а) {p_a}')
```

```python
p_b = round(9 / (9 * 10**4), 4)
print(f'б) {p_b}')
```

* Рассмотрим событие $C$ – все цифры нечетные

```python
p_с = round(5**5 / (9 * 10**4), 4)
print(f'в) {p_с}')
```

### Задача 1.48
> В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения 11 очков; б) выигрыша.

* Рассмотрим событие $A$ – сумма очков равна 11

```python
variants = list(itertools.product(range(1, 7), repeat=3))
total = len(variants)

sum_eq_11 = [el for el in variants if sum(el) == 11]
p_a = round(len(sum_eq_11) / total, 4)
print(f'а) {p_a}')
```

* Рассмотрим событие $B$ – выигрыш

```python
wins = [el for el in variants if sum(el) > 10]
p_b = round(len(wins) / total, 4)
print(f'б) {p_b}')
```
