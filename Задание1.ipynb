# Моделирование геометрической вероятности
Шаолиньский монах обучается мастерству кидания игральных карт, для этого он с закрытыми глазами кидает карты в воздушные шарики. Нужно смоделировать процесс кидания карт, если центры воздушных шариков образуют квадрат, гомотетичный стене с коэффициентом гомотетии равным 0,75 и центом гомотетии, совпадающим с центром стены, при этом площадь воздушного шарика в 9 раз меньше площади стены.

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Rectangle
import numpy as np
```

```python
def cards(size):
    R = 10
    side = np.sqrt(9 * np.pi * R ** 2)
    cord = [side/4, side*3/4]
    fig, ax = plt.subplots(figsize=(5, 5))
    ax = plt.gca()
    ax.add_patch(Rectangle((0, 0), side, side))
    ax.set_xlim(0, side)
    ax.set_ylim(0, side)
    for i in range(4):
        ax.add_patch(Circle((cord[i//2], cord[i%2]), R, facecolor='r'))

    X = np.random.uniform(0, side, size)
    Y = np.random.uniform(0, side, size)
    plt.scatter(X, Y, marker='x', c='black')

    inside = 0
    for i in range(4):
        inside += len([d for d in (X-cord[i//2])**2 + (Y-cord[i%2])**2 if d <= R**2])
    
    plt.show()
    print(f'В шарики из {size} кард попало {inside}. Доля попавших равна {inside/size}')
    
```

```python
cards(200)
```
